# Detecting and Converting Text Encodings to UTF-8
This notebook demonstrates how to inspect raw byte data, detect character encodings, and verify UTF-8 outputs produced by the conversion script.

## imports

```python
import os
from pathlib import Path
from charset_normalizer import detect
```

## config

```python
BASE_DIR = Path.cwd().parent
raw_dir = BASE_DIR / "data_raw"
out_dir = BASE_DIR / "data_utf8"
out_dir.mkdir(exist_ok=True)

files_names = sorted([p.name for p in raw_dir.iterdir()])
files_names
```

## read a file from disk as raw bytes.

```python
file_path = raw_dir / files_names[0]
if not file_path.is_file():
    raise FileNotFoundError(f"{file_path.name} is not a file")
        
data = file_path.read_bytes()
print(file_path.name, "bytes:", len(data))     
file_path
```

## guess the character encoding of raw byte data.

```python
result = detect(data)
enc = result.get("encoding")
print(result)
```

## decode raw bytes into a Unicode string using the given encoding.

```python
text = data.decode(enc, errors="strict")
print(text[:50])
```

## save decoded text to a UTF-8 encoded file.

```python
out_path = out_dir / f"utf8_{files_names[0]}"
out_path.write_text(text, encoding = "utf-8")
out_path
```

```python
assert (out_dir / "utf8_die_ISO-8859-1.txt").exists()
print("done")
```
